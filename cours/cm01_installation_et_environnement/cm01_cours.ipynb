### Python vs Java — même tâche, moins de code

````python
# Python : calculer une moyenne
notes = [12, 15, 18, 10, 14]
moyenne = sum(notes) / len(notes)
print(f"Moyenne : {moyenne}")
````

```java
// Java : même calcul — beaucoup plus verbeux
public class Moyenne {
    public static void main(String[] args) {
        int[] notes = {12, 15, 18, 10, 14};
        int somme = 0;
        for (int note : notes) { somme += note; }
        double moyenne = (double) somme / notes.length;
        System.out.println("Moyenne : " + moyenne);
    }
}
```

## 4. Git for Windows & terminal Bash

### 4.1 Pourquoi Git Bash ?

Sur Windows, les terminaux par défaut (**CMD** et **PowerShell**) posent des problèmes avec Python :
- les commandes Unix (`ls`, `cat`, `source`, `curl`) ne sont pas disponibles
- la gestion des chemins de fichiers et des variables d’environnement est différente
- la majorité de la documentation Python suppose un terminal Unix

**Git Bash** est un terminal Unix intégré dans **Git for Windows**. Il permet d’utiliser les mêmes commandes que sur Mac/Linux — et c’est le terminal que nous utiliserons tout au long de ce cours.

---

### 4.2 Installer Git for Windows

Télécharger depuis **[git-scm.com/download/win](https://git-scm.com/download/win)**

Le site détecte automatiquement votre système et propose la version 64-bit.

Pendant l’installation, **laisser toutes les options par défaut** — Git Bash est inclus automatiquement.

**Vérifier l’installation** — ouvrir Git Bash depuis le menu Démarrer :
```bash
git --version
# git version 2.x.x.windows.x
```

---

### 4.3 Configurer Git Bash comme terminal par défaut dans VS Code

Une fois Git for Windows installé, VS Code le détecte automatiquement :

1. Ouvrir VS Code
2. `Ctrl+Shift+P` → taper **`Terminal: Select Default Profile`**
3. Choisir **Git Bash**
4. Ouvrir un nouveau terminal avec `` Ctrl+` ``

Vous devriez voir `$` (Bash) au lieu de `>` (PowerShell).

**Vérifier que Bash est actif :**
```bash
echo $SHELL
# /usr/bin/bash
```

À partir de maintenant, tous vos terminaux VS Code s’ouvriront en Git Bash — les commandes `uv`, `python` et les outils Unix fonctionneront correctement.

## 6. Créer un projet Python avec uv

> **Démonstration live**

### 6.1 Qu’est-ce que uv ?

`uv` est le gestionnaire de projets Python moderne. Écrit en Rust, il est **10 à 100× plus rapide** que pip, et remplace en un seul outil :

| Ancien outil | Remplacé par uv |
|---|---|
| `python -m venv .venv` | automatique |
| `source .venv/bin/activate` | inutile |
| `pip install pandas` | `uv add pandas` |
| `pip freeze > requirements.txt` | `uv.lock` généré automatiquement |
| `pip install -r requirements.txt` | `uv sync` |

---

### 6.2 Installer uv

**Option 1 — Via Git Bash (recommandé) :**
```bash
curl -LsSf https://astral.sh/uv/install.sh | sh
```

**Option 2 — Via pip (si curl n’est pas disponible) :**
```bash
pip install uv
```

**Vérifier l’installation :**
```bash
uv --version
# uv 0.x.x
```

> Après l’installation via curl, redémarrer le terminal pour que la commande `uv` soit reconnue.

---

### 6.3 Créer un projet avec `uv init`

```bash
# Créer le dossier et initialiser le projet
uv init mon-projet
cd mon-projet

# Ouvrir dans VS Code
code .
```

**Convention de nommage** : utilisez des minuscules avec des tirets (`kebab-case`) — c’est le standard sur GitHub.

---

### 6.4 Structure du projet généré

`uv init` crée automatiquement :

```
mon-projet/
├── .git/                ← dépôt Git (uv init l'initialise)
├── .gitignore           ← ignore .venv/, __pycache__, etc.
├── .python-version      ← version Python du projet
├── pyproject.toml       ← configuration du projet et liste des dépendances
├── README.md            ← description du projet
└── main.py              ← fichier de démarrage
```

Le dossier `.venv/` et le fichier `uv.lock` (versions exactes des paquets)
apparaissent à la première commande qui a besoin de l'environnement :
`uv add`, `uv run` ou `uv sync`.

> Vous n’avez **jamais besoin d’activer** `.venv` — `uv` le gère automatiquement.

---

### 6.5 Ajouter des dépendances avec `uv add`

```bash
# Ajouter des paquets au projet
uv add pandas matplotlib numpy requests

# Exécuter un script dans l'environnement du projet
uv run hello.py

# Supprimer un paquet
uv remove requests
```

Les paquets sont automatiquement ajoutés dans `pyproject.toml` et leurs versions exactes sont enregistrées dans `uv.lock`.

---

### 6.6 Le fichier `pyproject.toml`

`pyproject.toml` est le fichier de configuration du projet. Il remplace `requirements.txt` :

```toml
[project]
name = "mon-projet"
version = "0.1.0"
requires-python = ">=3.12"
dependencies = [
    "pandas>=2.1.0",
    "matplotlib>=3.8.0",
    "numpy>=1.26.0",
]
```

---

### 6.7 `uv.lock` et `uv sync` — partager son environnement

`uv.lock` enregistre les **versions exactes** de tous les paquets installés (y compris leurs dépendances). Il garantit que tout le monde utilise exactement les mêmes versions.

```bash
# Sur un autre ordinateur : recréer l'environnement identique
uv sync
```

> Versionner `uv.lock` avec Git permet à toute l’équipe de travailler avec les mêmes dépendances, sans surprise.

---

### 6.8 `black` et `isort` — qualité de code

`black` et `isort` sont des outils qui **formatent automatiquement votre code** :

| Outil | Rôle |
|-------|------|
| **`black`** | Reformate le code Python selon un style standard universel |
| **`isort`** | Réorganise et trie les imports en tête de fichier |

Ce sont des dépendances de **développement** (utiles en local uniquement, pas en production) :

```bash
# Ajouter en tant que dépendances de développement
uv add --dev black isort

# Utiliser manuellement
uv run black mon_fichier.py
uv run isort mon_fichier.py
```

Dans `pyproject.toml`, elles apparaissent dans une section séparée :

```toml
[dependency-groups]
dev = [
    "black>=24.0.0",
    "isort>=5.13.0",
]
```

### 7.5 Comment Python exécute votre code

Python lit et exécute le code **de haut en bas, de gauche à droite**, ligne par ligne :

```
Ligne 1  →  Python lit  →  Python traduit  →  Exécute  →  Affiche
Ligne 2  →  même processus
Ligne 3  →  même processus
```

L’ordre compte : une ligne ne peut pas utiliser quelque chose défini plus bas.

````python
# L'ordre d'exécution est toujours de haut en bas
print("Étape 1 : Je commence")
x = 10
print(f"Étape 2 : x vaut {x}")
x = x * 2
print(f"Étape 3 : x vaut maintenant {x}")
print("Étape 4 : Fin du programme")
````

## 8. Paquets Python

### 8.1 Qu’est-ce qu’un paquet ?

Un paquet est du **code Python écrit par d’autres développeurs** que vous pouvez utiliser dans vos projets. Plutôt que de tout réécrire de zéro, vous réutilisez ce qui existe.

Exemples de paquets essentiels :

| Paquet | Utilisation |
|--------|------------|
| `requests` | Télécharger des données depuis le web |
| `pandas` | Manipuler des données tabulaires |
| `numpy` | Calculs mathématiques rapides |
| `matplotlib` | Créer des graphiques |
| `openai` | Connecter une application à ChatGPT |

Les paquets sont hébergés sur **PyPI** (Python Package Index) — une bibliothèque de plus de 500 000 paquets open-source.

### 8.2 Installer des paquets avec `uv add`

```bash
# Ajouter un paquet au projet
uv add requests

# Ajouter plusieurs paquets
uv add pandas matplotlib numpy

# Ajouter une version spécifique
uv add pandas==2.1.0

# Supprimer un paquet
uv remove requests
```

> `uv add` télécharge le paquet, met à jour `pyproject.toml` et `uv.lock` en une seule commande.

### 8.3 Utiliser un paquet installé

````python
# Après uv add requests, on peut l'importer et l'utiliser
import requests

reponse = requests.get("https://api.github.com")
print(f"Statut de l'API GitHub : {reponse.status_code}")
# 200 = succès
````

### 8.4 Méthodes d’import

````python
# Méthode 1 : importer le module entier
import math
print(math.pi)          # 3.14159...
print(math.sqrt(16))    # 4.0

# Méthode 2 : importer une fonction spécifique
from math import pi, sqrt
print(pi)               # 3.14159...
print(sqrt(25))         # 5.0

# Méthode 3 : importer avec un alias (convention pour pandas/numpy)
import numpy as np
import pandas as pd

tableau = np.array([1, 2, 3, 4, 5])
print(tableau * 2)      # [2 4 6 8 10]
````

## 9. Python Interactif — Le meilleur mode pour la data science

### 9.1 Pourquoi le mode interactif ?

Quand vous travaillez sur des données ou de l’IA, vous avez besoin de :
- Tester un bout de code sans relancer tout le fichier
- Voir les résultats immédiatement
- Garder les variables en mémoire d’une cellule à l’autre
- Expérimenter rapidement

Le mode interactif combine le meilleur des deux mondes :
- **Organisation** d’un fichier `.py` (pas de Jupyter Notebook désordonné)
- **Interactivité** de Jupyter (exécution cellule par cellule, résultats visuels)

### 9.2 Installer le kernel Jupyter

```bash
uv add ipykernel
```

### 9.3 Utiliser le mode interactif dans VS Code

- **Jupyter › Interactive Window › Text Editor: Execute Selection**

Dans un fichier `.py`, utilisez le raccourci **`Shift+Entrée`** :
- Sans sélection : exécute la ligne courante
- Avec une sélection : exécute le code sélectionné

Une fenêtre interactive s’ouvre à droite et affiche les résultats.

### 9.4 Délimiter des “cellules” dans un fichier `.py`

Vous pouvez structurer votre fichier en cellules avec `# %%` :

````python
# %% Cellule 1 : charger les données
import pandas as pd

data = {
    'nom': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'note': [15, 12, 18, 14],
    'ville': ['Amiens', 'Paris', 'Lyon', 'Amiens']
}
df = pd.DataFrame(data)
print(df)
````

````python
# %% Cellule 2 : analyser (df est toujours en mémoire)
print(f"Nombre d'étudiants : {len(df)}")
print(f"Note moyenne : {df['note'].mean():.1f}")
print(f"Meilleure note : {df['note'].max()}")
````

````python
# %% Cellule 3 : filtrer (toujours en mémoire)
amienois = df[df['ville'] == 'Amiens']
print("Étudiants d'Amiens :")
print(amienois)
````

## 10. Utiliser l’IA pour apprendre plus vite

Les assistants IA (ChatGPT, Claude) sont des outils très puissants pour apprendre Python. Voici comment les utiliser efficacement.

### Bons exemples de prompts

❌ **Mauvais** : *“Mon code ne marche pas, corrige-le”*

✅ **Bon** : *“J’apprends Python et j’essaie de calculer la moyenne d’une liste. Mon code : `sum(notes) / len(notes)`. Est-ce qu’il y a une façon plus efficace ?”*

### Stratégies efficaces

1. **Demander des explications** : “Explique-moi les fonctions Python simplement”
2. **Déboguer avec contexte** : coller le code ET le message d’erreur exact
3. **Demander des alternatives** : “Montre-moi 3 façons de résoudre ce problème”
4. **Vérifier la compréhension** : tester les suggestions et poser des questions de suivi

> L’IA est un outil pour **accélérer** l’apprentissage, pas pour le remplacer. L’objectif est de **comprendre**, pas d’avoir du code qui fonctionne sans savoir pourquoi.

## 11. Récapitulatif — Workflow complet

