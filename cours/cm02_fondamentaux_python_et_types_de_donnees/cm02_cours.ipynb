## 1. Qu'est-ce que la programmation ?

Programmer, c'est **écrire des instructions précises pour un ordinateur**, étape par étape — comme une recette de cuisine.

Les ordinateurs sont puissants mais pas intelligents : ils exécutent **exactement** ce qu'on leur dit, rien de plus.

Toute la programmation repose sur 4 opérations :

| Opération | Exemple Python |
|-----------|---------------|
| **Stocker de l'information** | `age = 25` |
| **Prendre des décisions** | `if age >= 18:` |
| **Répéter des actions** | `for i in range(10):` |
| **Calculer** | `total = prix * quantite` |

````python
# Premier aperçu de Python — lisible comme de l'anglais
nom = "Alice"
age = 20
print(f"Bonjour, je m'appelle {nom} et j'ai {age} ans.")

if age >= 18:
    print("Je suis majeur.")
else:
    print("Je suis mineur.")
````

## 2. Syntaxe Python & PEP8

### 2.1 L'indentation est obligatoire

La plupart des langages utilisent des accolades `{}`. Python utilise l'**indentation** (espaces en début de ligne) pour délimiter les blocs de code.

**Règle** : toujours **4 espaces** par niveau (jamais des tabulations).

````python
temperature = 35

if temperature > 30:
    print("Il fait chaud !")           # dans le bloc if — 4 espaces
    print("Pensez à vous hydrater.")   # toujours dans le bloc if

print("Fin du programme.")             # hors du bloc, toujours exécuté
````

````python
# Comparer Python et JavaScript pour le même code :

# Python
if temperature > 30:
    print("Chaud")

# JavaScript (pour référence — NE PAS exécuter)
# if (temperature > 30) {
#     console.log("Chaud");
# }
````

### 2.2 PEP8 — le guide de style officiel

**PEP8** est le document officiel qui définit les bonnes pratiques d'écriture Python. Le respecter, c'est écrire du code lisible par n'importe quel développeur Python.

| Règle | Mauvais | Bon |
|-------|---------|-----|
| 4 espaces | `\t` (tab) | `    ` (4 espaces) |
| Espaces autour de `=` | `x=5` | `x = 5` |
| Noms en snake_case | `firstName` | `first_name` |
| Lignes ≤ 79 caractères | ligne très longue | ligne découpée |
| Espaces après virgules | `[1,2,3]` | `[1, 2, 3]` |

````python
# Style non-PEP8 (fonctionne mais difficile à lire)
x=5;y=10;z=x+y;print(z)

print("---")

# Style PEP8 (professionnel, lisible)
x = 5
y = 10
z = x + y
print(z)
````

## 3. Comprendre et déboguer les erreurs

Les erreurs font partie de la programmation quotidienne. Python affiche toujours un message clair avec :
1. **Où** l'erreur se produit (numéro de ligne)
2. **Le type** d'erreur
3. **Un message** explicatif

**Méthode** : lire le message **de bas en haut** — la dernière ligne est la plus importante.

### Les 4 erreurs les plus fréquentes

````python
# SyntaxError : code invalide syntaxiquement
# Décommentez pour voir l'erreur :
# print("Bonjour"      # parenthèse non fermée
# if x > 5             # deux-points manquants
#     print(x)
````

````python
# NameError : variable utilisée avant d'être définie
# Décommentez pour voir l'erreur :
# print(salaire)   # NameError: name 'salaire' is not defined

# Solution : définir avant d'utiliser
salaire = 2500
print(salaire)
````

````python
# TypeError : mauvais type dans une opération
age = 25
# message = "J'ai " + age + " ans"   # TypeError !

# Solution 1 : convertir avec str()
message = "J'ai " + str(age) + " ans"
print(message)

# Solution 2 (recommandée) : f-string
message = f"J'ai {age} ans"
print(message)
````

````python
# IndentationError : indentation incorrecte
# Décommentez pour voir l'erreur :
# if True:
# print("Bonjour")   # IndentationError

# Correct
if True:
    print("Bonjour")
````

## 4. Variables

### 4.1 Créer une variable

Une variable est une **boîte étiquetée** en mémoire. On lui donne un nom et une valeur avec `=` (l'opérateur d'affectation — pas l'égalité mathématique).

````python
# Créer des variables
nom = "Alice"
age = 25
est_etudiant = True
taille = 1.68

print(nom, age, est_etudiant, taille)
````

````python
# Les variables peuvent changer
score = 0
print("Départ :", score)

score = 10
print("Après réponse 1 :", score)

score = score + 5   # la nouvelle valeur dépend de l'ancienne
print("Après bonus :", score)
````

### 4.2 Règles de nommage — snake_case

En Python, la convention est le **snake_case** : tout en minuscules, mots séparés par `_`.

| Valide | Invalide | Raison |
|--------|----------|--------|
| `first_name` | `first name` | espace interdit |
| `score_2024` | `2024_score` | ne peut pas commencer par un chiffre |
| `prix_ttc` | `prix-ttc` | tiret interdit (confusion avec soustraction) |
| `est_admin` | `class` | mot-clé réservé Python |

**Règle d'or** : choisissez des noms **descriptifs**. `age` vaut mieux que `a`, `prix_unitaire` vaut mieux que `p`.

````python
# Noms vagues vs noms descriptifs
# Mauvais
a = 80
b = 0.20
c = a * (1 + b)   # Que calcule-t-on ?

# Bon — le code se lit comme de l'anglais
prix_ht = 80
taux_tva = 0.20
prix_ttc = prix_ht * (1 + taux_tva)

print(f"Prix TTC : {prix_ttc} €")
````

## 6. Types de données

Chaque valeur en Python a un **type** qui détermine ce qu'on peut en faire.

| Type | Nom Python | Exemples |
|------|------------|----------|
| Entier | `int` | `42`, `-7`, `0` |
| Décimal | `float` | `3.14`, `19.99`, `-0.5` |
| Texte | `str` | `"Alice"`, `'Bonjour'` |
| Booléen | `bool` | `True`, `False` |

````python
# type() révèle le type d'une valeur
print(type(42))        # <class 'int'>
print(type(3.14))      # <class 'float'>
print(type("Alice"))   # <class 'str'>
print(type(True))      # <class 'bool'>
````

### 6.1 Nombres : `int` et `float`

- **`int`** : nombres entiers sans décimale
- **`float`** : nombres avec décimale (le point `.`, jamais la virgule)

````python
# Opérations arithmétiques
a, b = 10, 3

print(a + b)    # 13  — addition
print(a - b)    # 7   — soustraction
print(a * b)    # 30  — multiplication
print(a / b)    # 3.333... — division (TOUJOURS un float)
print(a // b)   # 3   — division entière (arrondit vers le bas)
print(a % b)    # 1   — modulo (reste)
print(a ** b)   # 1000 — puissance
````

````python
# PIÈGE : la division / retourne TOUJOURS un float
print(10 / 2)        # 5.0  (pas 5 !)
print(type(10 / 2))  # float

print(10 // 2)       # 5    (int)
print(type(10 // 2)) # int

# Nombres lisibles avec _ (convention Python)
population = 1_000_000   # plus lisible que 1000000
print(population)
````

````python
# Priorité des opérations (règles mathématiques : PEMDAS)
print(2 + 3 * 4)       # 14  — multiplication avant addition
print((2 + 3) * 4)     # 20  — parenthèses d'abord

# Exemple concret : calcul salaire annuel
salaire_mensuel = 2_500
primes = 1_200
salaire_annuel = salaire_mensuel * 12 + primes
print(f"Salaire annuel : {salaire_annuel} €")
````

### 6.2 Chaînes de caractères (`str`)

Du texte entre guillemets simples `'...'` ou doubles `"..."` — Python accepte les deux.

````python
# Créer des strings
prenom = "Alice"
ville = 'Amiens'

# Multilignes avec triple guillemets
description = """Ce cours couvre
Python pour la data science
à l'UPJV Amiens."""

print(prenom, "-", ville)
print(description)
````

````python
# Concaténation avec +
prenom = "John"
nom = "Doe"
nom_complet = prenom + " " + nom
print(nom_complet)     # John Doe

# Répétition avec *
separateur = "-" * 40
print(separateur)

# Longueur avec len()
message = "Bonjour le monde"
print(f"'{message}' contient {len(message)} caractères")
````

### 6.3 F-strings — insérer des variables dans du texte

Préfixer la chaîne avec `f` et mettre les variables entre `{ }`. C'est la méthode moderne et recommandée.

````python
prenom = "Alice"
age = 20
note = 14.75

# Ancienne méthode — difficile à lire
message = "Bonjour " + prenom + ", tu as " + str(age) + " ans."
print(message)

# F-string — lisible et moderne
message = f"Bonjour {prenom}, tu as {age} ans."
print(message)

# Expressions dans {}
prix_ht = 80
taux_tva = 0.20
print(f"Prix TTC : {prix_ht * (1 + taux_tva):.2f} €")  # :.2f = 2 décimales
print(f"Note : {note:.1f}/20")   # 1 décimale
````

### 6.4 Méthodes des chaînes

Python fournit plus de 40 méthodes pour manipuler les strings. Les plus utiles :

````python
texte = "  Python Data Science  "

# Casse
print(texte.lower())     # tout en minuscules
print(texte.upper())     # tout en majuscules
print(texte.title())     # première lettre de chaque mot en majuscule

# Nettoyage des espaces
print(texte.strip())     # supprime espaces de début et fin

# Recherche et remplacement
message = "J'aime Python et Python c'est super"
print("Python" in message)                    # True
print(message.count("Python"))                # 2
print(message.replace("Python", "Python 3"))  # remplace toutes les occurrences
````

````python
# split() : découper une chaîne → liste
csv_ligne = "Économie,Finance,Statistiques,Python"
matieres = csv_ligne.split(",")
print(matieres)        # ['Économie', 'Finance', 'Statistiques', 'Python']
print(len(matieres))   # 4

# join() : assembler une liste → chaîne
resultat = " | ".join(matieres)
print(resultat)        # Économie | Finance | Statistiques | Python

# Chaîner les méthodes
entree_utilisateur = "  ALICE  "
propre = entree_utilisateur.strip().lower().title()
print(propre)          # Alice
````

### 6.5 Booléens (`bool`)

Seulement deux valeurs possibles : **`True`** ou **`False`** (majuscule obligatoire !).

````python
# Créer des booléens directement
est_connecte = True
est_admin = False

print(type(est_connecte))   # <class 'bool'>

# Les booléens viennent souvent de comparaisons
age = 20
note = 55

peut_voter = age >= 18         # True
a_valide = note >= 60          # False

print(f"Peut voter : {peut_voter}")
print(f"A validé : {a_valide}")
````

### 6.6 Conversions de types

````python
# int() — convertir en entier (tronque, n'arrondit pas)
print(int(3.9))      # 3  (pas 4 !)
print(int("42"))     # 42
print(int(True))     # 1

# float() — convertir en décimal
print(float(5))      # 5.0
print(float("3.14")) # 3.14

# str() — convertir en texte
print(str(42))       # "42"
print(str(True))     # "True"

# bool() — valeurs falsy : 0, 0.0, "", [], {}
print(bool(0))       # False
print(bool(""))      # False
print(bool("hello")) # True
print(bool(-5))      # True (tout entier non nul)
````

## 7. Opérateurs

### 7.1 Opérateurs de comparaison

Comparent deux valeurs et retournent un booléen.

| Opérateur | Signification | Exemple |
|-----------|---------------|---------|
| `==` | égal à | `5 == 5` → `True` |
| `!=` | différent de | `5 != 3` → `True` |
| `>` | supérieur strict | `7 > 3` → `True` |
| `<` | inférieur strict | `2 < 1` → `False` |
| `>=` | supérieur ou égal | `5 >= 5` → `True` |
| `<=` | inférieur ou égal | `3 <= 4` → `True` |

````python
age = 18
print(age == 18)   # True
print(age != 21)   # True
print(age > 17)    # True
print(age < 18)    # False
print(age >= 18)   # True

# PIÈGE CLASSIQUE : = vs ==
x = 10          # affectation : stocke 10 dans x
print(x == 10)  # comparaison : True
print(x == 5)   # comparaison : False
````

### 7.2 Opérateurs logiques

| Opérateur | Comportement |
|-----------|--------------|
| `and` | Vrai si **les deux** conditions sont vraies |
| `or` | Vrai si **au moins une** condition est vraie |
| `not` | **Inverse** la valeur booléenne |

````python
age = 25
a_permis = True

# AND — les deux doivent être True
peut_conduire = age >= 18 and a_permis
print(f"Peut conduire : {peut_conduire}")   # True

# OR — au moins un doit être True
jour = "Samedi"
est_conge = jour == "Samedi" or jour == "Dimanche"
print(f"Jour de congé : {est_conge}")       # True

# NOT — inverse
est_adulte = age >= 18
est_mineur = not est_adulte
print(f"Est mineur : {est_mineur}")          # False
````

````python
# Tables de vérité
print("AND:", True and True, True and False, False and False)
print("OR: ", True or False, False or False)
print("NOT:", not True, not False)
````

### 7.3 Raccourcis d'affectation

````python
score = 0

score = score + 10   # forme longue
print(score)         # 10

score += 5           # équivalent à score = score + 5
print(score)         # 15

score *= 2           # score = score * 2
print(score)         # 30

score -= 8           # score = score - 8
print(score)         # 22

score //= 3          # score = score // 3
print(score)         # 7
````
