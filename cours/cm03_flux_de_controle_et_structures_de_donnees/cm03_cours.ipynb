## 1. Flux de contrôle

Jusqu'ici, vos programmes s'exécutent de haut en bas, ligne par ligne. Le flux de contrôle permet de **prendre des décisions** et **répéter des actions** — c'est ce qui rend les programmes intelligents.

Exemples du quotidien :
- **SI** le mot de passe est correct → accès autorisé

## 2. Instructions conditionnelles (`if / elif / else`)

### 2.1 `if` de base

````python
age = 18

if age >= 18:
    print("Vous pouvez voter !")      # exécuté si True
    print("Vous êtes majeur.")

print("Programme terminé.")           # toujours exécuté
````

Le `:` et l'indentation (4 espaces) sont **obligatoires** — c'est ainsi que Python délimite le bloc conditionnel.

### 2.2 `if / else`

````python
temperature = 22

if temperature > 30:
    print("Il fait chaud !")
else:
    print("La météo est agréable.")
````

### 2.3 `if / elif / else` — plusieurs conditions

Python vérifie les conditions **dans l'ordre** et exécute le premier bloc `True`.

````python
note = 85

if note >= 90:
    mention = "Très bien"
elif note >= 80:
    mention = "Bien"
elif note >= 70:
    mention = "Assez bien"
elif note >= 60:
    mention = "Passable"
else:
    mention = "Insuffisant"

print(f"Note : {note}/100 — {mention}")
````

````python
# Combiner des conditions avec and / or / not
age = 25
a_permis = True

if age >= 18 and a_permis:
    print("Peut conduire")

jour = "Samedi"
if jour == "Samedi" or jour == "Dimanche":
    print("Week-end !")

# Conditions imbriquées
a_billet = True
age_cinema = 15

if a_billet:
    if age_cinema >= 18:
        print("Film autorisé")
    else:
        print("Film interdit aux mineurs — accompagnement requis")
else:
    print("Achetez d'abord un billet")
````

### 2.4 Erreurs fréquentes avec les conditions

````python
x = 10

# ERREUR 1 : oublier les deux-points
# if x > 5    # SyntaxError !

# ERREUR 2 : = au lieu de ==
# if x = 5:   # SyntaxError !

# ERREUR 3 : mauvaise indentation
# if x > 5:
# print("Grand")  # IndentationError !

# CORRECT
if x > 5:
    print(f"{x} est grand")
elif x == 5:
    print("Exactement 5")
else:
    print(f"{x} est petit")
````

## 3. Boucles

Les boucles permettent de **répéter du code** sans le copier-coller.

### 3.1 La boucle `for`

La boucle `for` est la plus utilisée en Python — elle parcourt une séquence (range, liste, string…).

````python
# Répéter N fois avec range()
for i in range(5):
    print(i)   # 0, 1, 2, 3, 4

# Python compte à partir de 0 — c'est le "zero-indexing"
````

````python
# range() : 3 formes
print("range(5)      :", list(range(5)))       # 0 à 4
print("range(1, 6)   :", list(range(1, 6)))    # 1 à 5
print("range(0,10,2) :", list(range(0, 10, 2))) # 0, 2, 4, 6, 8 (pas de 2)
````

````python
# Parcourir un string lettre par lettre
mot = "Python"
for lettre in mot:
    print(lettre, end=" ")   # end=" " évite le retour à la ligne
print()  # retour à la ligne final
````

````python
# Exemple concret : calculer la somme des notes
notes = [12, 15, 18, 10, 14]
somme = 0

for note in notes:
    somme += note

moyenne = somme / len(notes)
print(f"Somme : {somme}  |  Moyenne : {moyenne:.1f}")
````

### 3.2 La boucle `while`

Continue **tant qu'une condition est vraie**. Toujours s'assurer qu'elle finit par s'arrêter !

````python
compte = 0
while compte < 5:
    print(f"Compte : {compte}")
    compte += 1   # sans cette ligne → boucle infinie !
````

````python
# Exemple : trouver le premier multiple de 7 supérieur à 100
n = 101
while n % 7 != 0:
    n += 1
print(f"Premier multiple de 7 > 100 : {n}")
````

### 3.3 Erreurs fréquentes avec les boucles

````python
# ERREUR 1 : off-by-one — range(5) donne 0,1,2,3,4 (pas 5)
print("Étudiants 1 à 5 :")
for i in range(1, 6):   # range(1, 6) = 1, 2, 3, 4, 5
    print(f"  Étudiant {i}")

# ERREUR 2 : oublier d'incrémenter dans while (boucle infinie)
# compteur = 0
# while compteur < 5:        # Danger !
#     print(compteur)        # compteur ne change jamais

# ERREUR 3 : oublier les deux-points
# for i in range(5)    # SyntaxError

print("OK")
````

## 4. Structures de données

Jusqu'ici, chaque variable stockait **une seule valeur**. Les structures de données permettent d'en stocker **plusieurs**.

| Structure | Analogie | Ordonné | Modifiable | Unique |
|-----------|----------|---------|-----------|--------|
| **Liste** | Liste de courses | ✅ | ✅ | ❌ |
| **Dictionnaire** | Annuaire | ✅ (ordre d'insertion, depuis Python 3.7) | ✅ | Clés uniquement |
| **Tuple** | Coordonnées GPS | ✅ | ❌ | ❌ |
| **Ensemble** | Sac de billes uniques | ❌ | ✅ | ✅ |

> Python compte à partir de **0** — le premier élément est à l'index 0, le second à 1, etc. C'est le "zero-based indexing".

---
### 4.1 Listes (`list`)

La structure la plus utilisée en Python. **Ordonnée, modifiable, accepte les doublons.**

````python
# Créer des listes
fruits = ["pomme", "banane", "orange"]
notes = [12, 15, 18, 10, 14]
mixte = ["Alice", 20, True, 3.14]    # types différents OK
vide = []

print(fruits)
print(type(fruits))   # <class 'list'>
````

````python
# Accéder aux éléments par index
fruits = ["pomme", "banane", "orange", "kiwi"]

print(fruits[0])    # "pomme"  — premier
print(fruits[1])    # "banane"
print(fruits[-1])   # "kiwi"   — dernier
print(fruits[-2])   # "orange" — avant-dernier

# Slicing : extraire une sous-liste
print(fruits[0:2])  # ['pomme', 'banane']  — indices 0 et 1
print(fruits[1:])   # ['banane', 'orange', 'kiwi']
print(fruits[:2])   # ['pomme', 'banane']
````

````python
# Modifier une liste
fruits = ["pomme", "banane", "orange"]

fruits[0] = "mangue"         # modifier un élément
fruits.append("raisin")      # ajouter à la fin
fruits.insert(1, "kiwi")     # insérer à la position 1
print(fruits)

fruits.remove("banane")      # supprimer par valeur
dernier = fruits.pop()       # supprimer et retourner le dernier
print(f"Retiré : {dernier}")
print(fruits)
````

````python
# Méthodes utiles
nombres = [3, 1, 4, 1, 5, 9, 2, 6]

print("Longueur :", len(nombres))
print("Min / Max :", min(nombres), "/", max(nombres))
print("Somme :", sum(nombres))
print("Occurrences de 1 :", nombres.count(1))

nombres.sort()
print("Triée :", nombres)

nombres.reverse()
print("Inversée :", nombres)

# Vérifier la présence
print("5 dans la liste :", 5 in nombres)
````

````python
# Parcourir une liste avec for
etudiants = ["Alice", "Bob", "Charlie", "Diana"]

for etudiant in etudiants:
    print(f"Bonjour {etudiant} !")

print()

# enumerate() pour avoir l'index ET la valeur
for i, etudiant in enumerate(etudiants, start=1):
    print(f"{i}. {etudiant}")
````

### 4.2 Dictionnaires (`dict`)

Stockent des données sous forme **clé → valeur**. Comme un annuaire : on cherche par le nom (clé) pour trouver le numéro (valeur).

````python
# Créer un dictionnaire
etudiant = {
    "nom": "Alice",
    "age": 20,
    "ville": "Amiens",
    "note": 15.5
}

print(etudiant)
print(type(etudiant))   # <class 'dict'>
````

````python
# Accéder aux valeurs
print(etudiant["nom"])           # "Alice"
print(etudiant["note"])          # 15.5

# Méthode .get() — sécurisée (pas d'erreur si clé manquante)
print(etudiant.get("email"))              # None
print(etudiant.get("email", "inconnu"))   # "inconnu" (valeur par défaut)
````

````python
# Modifier un dictionnaire
etudiant["email"] = "alice@upjv.fr"   # ajouter une nouvelle clé
etudiant["note"] = 16.0               # modifier une valeur existante
del etudiant["ville"]                  # supprimer une clé
print(etudiant)
````

````python
# Parcourir un dictionnaire
etudiant = {"nom": "Alice", "age": 20, "note": 16.0, "email": "alice@upjv.fr"}

# Clés
print("Clés :", list(etudiant.keys()))

# Valeurs
print("Valeurs :", list(etudiant.values()))

# Paires clé-valeur
for cle, valeur in etudiant.items():
    print(f"  {cle} : {valeur}")
````

````python
# Dictionnaire imbriqué — très courant en data science
etudiants = {
    "alice": {"age": 20, "note": 15.5, "mention": "Bien"},
    "bob": {"age": 21, "note": 12.0, "mention": "Passable"},
    "charlie": {"age": 19, "note": 18.0, "mention": "Très bien"}
}

# Accéder à une donnée précise
print(etudiants["alice"]["mention"])   # "Bien"

# Afficher le tableau de notes
print(f"{'Nom':<12} {'Note':>6} {'Mention'}")
print("-" * 30)
for nom, info in etudiants.items():
    print(f"{nom.title():<12} {info['note']:>6.1f} {info['mention']}")
````

### 4.3 Tuples (`tuple`)

Comme une liste, mais **immuable** (impossible à modifier après création). Utilisés pour des données qui ne doivent pas changer : coordonnées GPS, couleurs RGB, valeurs de retour de fonctions.

````python
# Créer des tuples
point = (3, 5)                        # coordonnées
couleur_rouge = (255, 0, 0)           # RGB
mois = ("Jan", "Fév", "Mar", "Avr")  # liste fixe

# Un seul élément : virgule obligatoire
seul = (42,)          # tuple
pas_tuple = (42)      # juste le nombre 42 !
print(type(seul), type(pas_tuple))
````

````python
# Accès par index — comme les listes
point = (3, 5)
print(point[0])   # 3
print(point[-1])  # 5

# Tuple unpacking — très élégant
x, y = point
print(f"x={x}, y={y}")

# Échanger deux variables sans variable temporaire
a, b = 10, 20
a, b = b, a
print(f"a={a}, b={b}")   # a=20, b=10

# Affectations multiples
latitude, longitude, altitude = 49.894, 2.296, 34
print(f"Amiens : {latitude}°N, {longitude}°E, {altitude}m")
````

````python
# Immuable = impossible à modifier
point = (3, 5)
# point[0] = 4   # TypeError : 'tuple' object does not support item assignment

# Pourquoi utiliser un tuple plutôt qu'une liste ?
# 1. Signale que la donnée ne doit pas changer
# 2. Plus rapide et moins gourmand en mémoire
# 3. Peut servir de clé dans un dictionnaire (les listes ne peuvent pas)

positions = {(0, 0): "origine", (1, 0): "droite", (0, 1): "haut"}
print(positions[(0, 0)])
````

### 4.4 Ensembles (`set`)

Collections de valeurs **uniques** sans ordre défini. Parfaits pour éliminer les doublons et faire des tests d'appartenance rapides.

````python
# Créer un ensemble
langages = {"Python", "R", "SQL", "Python"}  # Python apparaît 2 fois
print(langages)   # Python n'apparaît qu'une fois — les doublons sont supprimés

# ATTENTION : {} seul crée un dict, pas un set !
vide_dict = {}       # dict
vide_set = set()     # set — toujours utiliser set()
print(type(vide_dict), type(vide_set))
````

````python
# Cas d'usage 1 : supprimer les doublons d'une liste
reponses = ["Python", "R", "Python", "SQL", "R", "Python"]
print("Avec doublons :", reponses)

uniques = list(set(reponses))
print("Sans doublons :", sorted(uniques))   # sorted pour affichage stable
````

````python
# Cas d'usage 2 : test d'appartenance (très rapide)
utilisateurs_autorises = {"alice", "bob", "charlie", "diana"}

demandeur = "eve"
if demandeur in utilisateurs_autorises:
    print(f"{demandeur} : accès autorisé")
else:
    print(f"{demandeur} : accès refusé")
````

````python
# Opérations ensemblistes
matieres_l3 = {"Python", "Statistiques", "Économétrie", "Finance"}
matieres_m1 = {"Python", "Machine Learning", "Économétrie", "Big Data"}

# Intersection : matières communes
communes = matieres_l3 & matieres_m1
print("Communes :", communes)

# Union : toutes les matières
toutes = matieres_l3 | matieres_m1
print("Toutes :", toutes)

# Différence : matières en L3 seulement
l3_seulement = matieres_l3 - matieres_m1
print("L3 uniquement :", l3_seulement)
````

## 5. Choisir la bonne structure

| Situation | Structure |
|-----------|-----------|
| Liste ordonnée de valeurs qui peuvent changer | `list` |
| Associer des noms à des valeurs | `dict` |
| Données fixes qui ne doivent pas changer | `tuple` |
| Valeurs uniques, test d'appartenance rapide | `set` |

````python
# Exemple complet : profil d'un étudiant
profil = {
    "nom": "Alice Dupont",
    "notes": [14, 16, 12, 18, 15],       # liste : ordonnée, modifiable
    "diplomes": ("Bac S", "Licence"),    # tuple : figé
    "langages": {"Python", "R", "SQL"}   # set : sans doublons
}

# Calculs
moyenne = sum(profil["notes"]) / len(profil["notes"])
print(f"Étudiant : {profil['nom']}")
print(f"Moyenne : {moyenne:.1f}/20")
print(f"Diplômes : {', '.join(profil['diplomes'])}")
print(f"Maîtrise Python : {'Python' in profil['langages']}")
````
